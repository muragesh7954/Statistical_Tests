**Variance for returns**

```python
import yfinance as yf
import pandas as pd
import datetime

# Define the ticker symbol for TCS
ticker_symbol = "TCS.NS" # .NS for National Stock Exchange (India)

# Define the end date as today and start date as one year ago
end_date = datetime.date.today()
start_date = end_date - datetime.timedelta(days=365)

# Download historical data
df = yf.download(ticker_symbol, start=start_date, end=end_date)

# Calculate daily returns
df["Daily_Return"] = df["Close"].pct_change()

print(f"Downloaded {len(df)} days of data for {ticker_symbol} from {start_date} to {end_date}.")
print("First 5 rows of the DataFrame with Daily_Return:")
print(df.head())
```

```python
import numpy as np
import pandas as pd

# Example:
# df["Daily_Return"] = df["Close"].pct_change()

# Remove missing values
returns = df["Daily_Return"].dropna().to_numpy()

# ----------------------------------------------------
# Sample Variance (Recommended for historical returns)
# ddof = 1 -> divides by (n-1)
# ----------------------------------------------------
variance = np.var(returns, ddof=1)

print("=" * 60)
print(f"Variance : {variance:.8f}")
print("=" * 60)

# ----------------------------------------------------
# Volatility Decision
# ----------------------------------------------------

if variance < 0.0001:
    print("Volatility Level : Very Low")
    print("Risk             : Very Low")
    print("Interpretation   : Returns are highly stable.")
    print("Decision         : Suitable for conservative investors.")

elif variance < 0.0004:
    print("Volatility Level : Low")
    print("Risk             : Low")
    print("Interpretation   : Small fluctuations in returns.")
    print("Decision         : Suitable for long-term investing.")

elif variance < 0.0016:
    print("Volatility Level : Moderate")
    print("Risk             : Medium")
    print("Interpretation   : Normal market volatility.")
    print("Decision         : Diversification is recommended.")

elif variance < 0.0036:
    print("Volatility Level : High")
    print("Risk             : High")
    print("Interpretation   : Large price fluctuations.")
    print("Decision         : Use stop-loss and proper position sizing.")

else:
    print("Volatility Level : Very High")
    print("Risk             : Very High")
    print("Interpretation   : Extreme market volatility.")
    print("Decision         :")
    print("  • Reduce portfolio exposure")
    print("  • Hedge with Options/Futures")
    print("  • Perform stress testing")
    print("  • Monitor positions frequently")
```
