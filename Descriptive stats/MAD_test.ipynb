```python
import numpy as np
import pandas as pd

# ----------------------------------------------------------
# Example:
# df["Daily_Return"] = df["Close"].pct_change()
# Daily_Return should be in decimal format
# Example: 0.012 = 1.2%
# ----------------------------------------------------------

# Remove missing values
returns = df["Daily_Return"].dropna().to_numpy()

# ----------------------------------------------------------
# Standard Deviation (Sample)
# ----------------------------------------------------------
std_dev = np.std(returns, ddof=1)

# ----------------------------------------------------------
# Median Absolute Deviation (MAD)
# ----------------------------------------------------------
median = np.median(returns)

mad = np.median(np.abs(returns - median))

print("=" * 70)
print(f"Median Return              : {median:.4%}")
print(f"Median Absolute Deviation  : {mad:.4%}")
print(f"Standard Deviation         : {std_dev:.4%}")
print("=" * 70)

# ----------------------------------------------------------
# MAD Decision
# ----------------------------------------------------------
print("\nVOLATILITY ANALYSIS USING MAD")
print("-" * 70)

if mad < 0.005:
    print("MAD Level        : Very Low")
    print("Interpretation   : Returns stay very close to the median.")
    print("Decision         : Stable portfolio with low day-to-day variability.")

elif mad < 0.015:
    print("MAD Level        : Moderate")
    print("Interpretation   : Normal fluctuations around the median.")
    print("Decision         : Suitable for balanced portfolios.")

elif mad < 0.030:
    print("MAD Level        : High")
    print("Interpretation   : Returns fluctuate noticeably around the median.")
    print("Decision         : Monitor risk and position sizing.")

else:
    print("MAD Level        : Very High")
    print("Interpretation   : Large day-to-day fluctuations.")
    print("Decision         : Consider diversification or hedging.")

# ----------------------------------------------------------
# Compare MAD with Standard Deviation
# ----------------------------------------------------------
print("\nCOMPARISON WITH STANDARD DEVIATION")
print("-" * 70)

ratio = mad / std_dev if std_dev != 0 else 0

print(f"MAD / Standard Deviation Ratio : {ratio:.2f}")

if ratio > 0.75:
    print("Interpretation :")
    print("• MAD and Standard Deviation are similar.")
    print("• Few extreme outliers.")
    print("• Volatility is driven mainly by typical daily movements.")

elif ratio > 0.50:
    print("Interpretation :")
    print("• Moderate difference between MAD and Standard Deviation.")
    print("• Some unusually large returns may exist.")
    print("• Review for occasional market shocks.")

else:
    print("Interpretation :")
    print("• Standard Deviation is much larger than MAD.")
    print("• A few extreme returns are inflating volatility.")
    print("• Investigate outliers or rare market events.")
    print("• Consider additional tail-risk metrics (Skewness, Kurtosis, VaR).")
```
