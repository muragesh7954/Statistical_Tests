```python
import numpy as np
import pandas as pd

# ----------------------------------------------------
# Example:
# df["Daily_Return"] = df["Close"].pct_change()
# Returns should be decimal values
# Example:
# 0.01 = 1%
# ----------------------------------------------------

returns = df["Daily_Return"].dropna().to_numpy()

# ----------------------------------------------------
# Statistics
# ----------------------------------------------------

mean_return = np.mean(returns)
std_return = np.std(returns, ddof=1)

print("=" * 70)
print(f"Mean Return          : {mean_return:.4%}")
print(f"Standard Deviation   : {std_return:.4%}")
print("=" * 70)

# ----------------------------------------------------
# Number of Standard Deviations
# ----------------------------------------------------

k = 2      # Change to 3,4,5 if required

# ----------------------------------------------------
# Chebyshev Lower Bound
# ----------------------------------------------------

lower_bound = 1 - (1 / (k ** 2))

# ----------------------------------------------------
# Confidence Interval
# ----------------------------------------------------

lower = mean_return - k * std_return
upper = mean_return + k * std_return

print(f"\nChebyshev Interval ({k}σ)")
print("-" * 70)
print(f"Lower Limit : {lower:.4%}")
print(f"Upper Limit : {upper:.4%}")
print(f"Theoretical Minimum Coverage : {lower_bound:.2%}")

# ----------------------------------------------------
# Actual Coverage
# ----------------------------------------------------

inside = np.sum((returns >= lower) & (returns <= upper))

coverage = inside / len(returns)

print(f"Actual Coverage             : {coverage:.2%}")

# ----------------------------------------------------
# Decision Making
# ----------------------------------------------------

print("\nPORTFOLIO RISK ANALYSIS")
print("-" * 70)

if coverage >= lower_bound:

    print("Status           : PASS")
    print("Interpretation   :")
    print("Portfolio satisfies Chebyshev's inequality.")

    if coverage > 0.98:
        print("Risk Level       : Very Low")
        print("Decision         :")
        print("- Stable return distribution.")
        print("- Suitable for long-term investors.")

    elif coverage > 0.95:
        print("Risk Level       : Low")
        print("Decision         :")
        print("- Normal market volatility.")
        print("- Portfolio is reasonably stable.")

    elif coverage > 0.90:
        print("Risk Level       : Moderate")
        print("Decision         :")
        print("- Monitor market conditions.")
        print("- Diversification recommended.")

    else:
        print("Risk Level       : Elevated")
        print("Decision         :")
        print("- Returns are more dispersed.")
        print("- Review asset allocation.")

else:

    print("Status           : FAIL")
    print("Interpretation   :")
    print("Observed coverage is below the theoretical guarantee.")
    print("This may indicate data issues, numerical problems, or an incorrect implementation.")
    print("Chebyshev's inequality should hold for data with finite mean and variance.")

print("=" * 70)
```
