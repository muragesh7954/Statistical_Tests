**Compute percentiles**

```python
import numpy as np
import pandas as pd
```

```python
import yfinance as yf
from datetime import datetime, timedelta

ticker_symbol = "TCS.NS"  # Tata Consultancy Services on NSE

# Define the end date as today and start date as one year ago
end_date = datetime.now()
start_date = end_date - timedelta(days=365)

# Download hourly data for the specified period
tcs_data = yf.download(ticker_symbol, start=start_date, end=end_date, interval="1h")

# Display the first few rows of the downloaded data
print(f"Downloaded {len(tcs_data)} rows of hourly data for {ticker_symbol}")
print(tcs_data.head())
```

```python
#calculating 0-100th percentile to find a the correct percentile value for removal of outliers
for i in range(0,100,10):
    var =tcs_data["Volume"].values
    #print(var)
    var = np.sort(var,axis = None)
    print("{} percentile value is {}".format(i,var[int(len(var)*(float(i)/100))]))
print ("100 percentile value is ",var[-1])
```

```python
#looking further from the 99th percecntile
for i in range(90,100):
    var =tcs_data["Volume"].values
    var = np.sort(var,axis = None)
    print("{} percentile value is {}".format(i,var[int(len(var)*(float(i)/100))]))
print ("100 percentile value is ",var[-1])
```

```python
#calculating sped values at each percntile 99.0,99.1,99.2,99.3,99.4,99.5,99.6,99.7,99.8,99.9,100
for i in np.arange(0.0, 1.0, 0.1):
    var =tcs_data["Volume"].values
    var = np.sort(var,axis = None)
    print("{} percentile value is {}".format(99+i,var[int(len(var)*(float(99+i)/100))]))
print("100 percentile value is ",var[-1])
```
