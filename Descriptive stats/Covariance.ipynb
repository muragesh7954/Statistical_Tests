```python
import numpy as np
import pandas as pd

# Example daily returns of two portfolios
portfolio_returns = pd.DataFrame({
    "Portfolio_A": [0.010, 0.015, -0.005, 0.008, 0.012,
                   -0.010, 0.006, 0.009, -0.004, 0.011],

    "Portfolio_B": [0.008, 0.012, -0.002, 0.006, 0.010,
                   -0.008, 0.004, 0.007, -0.003, 0.009]
})

# Compute sample covariance
covariance = portfolio_returns["Portfolio_A"].cov(
    portfolio_returns["Portfolio_B"]
)

print("Covariance:", covariance)


# Make portfolio decision
if covariance > 0:
    decision = "Positive covariance: Portfolios tend to move in the same direction."
    recommendation = "Limited diversification benefit. Consider adding less-correlated assets."

elif covariance < 0:
    decision = "Negative covariance: Portfolios tend to move in opposite directions."
    recommendation = "Good diversification benefit. Holding both portfolios can reduce portfolio risk."

else:
    decision = "Covariance is approximately zero: No linear co-movement."
    recommendation = "Potential diversification benefit because the portfolios are largely independent."

print("\nDecision:", decision)
print("Recommendation:", recommendation)
```
