**Z Score for returns**

```python
import pandas as pd
from scipy.stats import zscore

returns = df["Daily_Return"]

df["Return_Z"] = zscore(
    returns,
    ddof=1
)

print(df.head())
```

```python
latest = df["Return_Z"].iloc[-1]

print(f"Today's Z-score : {latest:.2f}")

if abs(latest) < 2:
    print("Normal Market Movement")

elif 2 <= abs(latest) < 3:
    print("Unusual Price Movement")
    print("Monitor the stock carefully.")

else:
    print("Potential Outlier")
    print("Possible news event or market shock.")
```

**Outlier Detection**

```python
import pandas as pd
from scipy.stats import zscore

df["Z"] = zscore(df["Daily_Return"])

outliers = df[abs(df["Z"]) >= 3]

print(outliers)
```
