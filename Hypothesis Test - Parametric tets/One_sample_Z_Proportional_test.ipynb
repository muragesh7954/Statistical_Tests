one-sample proportion Z-test using a trading-strategy example. We will test whether the strategy's win rate is greater than 50%, and make the decision using both the p-value and critical Z-value.

Example

Suppose:

Total trades = 200
Profitable trades = 130
Observed win rate = 130/200=65%
Benchmark win rate = 50%
Significance level = 5%
Alternative = greater

Hypotheses:
H0​:p≤0.50
H1​:p>0.50

```python
import numpy as np
from scipy.stats import norm


# ============================================================
# 1. Sample information
# ============================================================

total_trades = 200
profitable_trades = 130


# ============================================================
# 2. Hypothesized population proportion
# ============================================================

# Benchmark / target win rate
p_0 = 0.50


# ============================================================
# 3. Significance level
# ============================================================

alpha = 0.05


# ============================================================
# 4. Calculate sample proportion
# ============================================================

p_hat = profitable_trades / total_trades


# ============================================================
# 5. Check assumptions
# ============================================================

expected_successes = total_trades * p_0
expected_failures = total_trades * (1 - p_0)

if expected_successes < 5 or expected_failures < 5:
    print("Warning: Normal approximation may not be appropriate.")


# ============================================================
# 6. Calculate standard error
# ============================================================

standard_error = np.sqrt(
    (p_0 * (1 - p_0)) / total_trades
)


# ============================================================
# 7. Calculate Z-statistic
# ============================================================

z_stat = (p_hat - p_0) / standard_error


# ============================================================
# 8. Right-tailed p-value
#
# H1: p > p0
# ============================================================

p_value = norm.sf(z_stat)


# ============================================================
# 9. Critical Z-value
# ============================================================

z_critical = norm.ppf(1 - alpha)


# ============================================================
# 10. Decision using p-value
# ============================================================

if p_value < alpha:
    decision_p = "Reject H0"
    conclusion_p = (
        "There is statistically significant evidence that "
        "the strategy's win rate is greater than 50%."
    )
else:
    decision_p = "Fail to reject H0"
    conclusion_p = (
        "There is insufficient statistical evidence to conclude "
        "that the strategy's win rate is greater than 50%."
    )


# ============================================================
# 11. Decision using critical Z-value
# ============================================================

if z_stat > z_critical:
    decision_z = "Reject H0"
else:
    decision_z = "Fail to reject H0"


# ============================================================
# 12. Display results
# ============================================================

print("=" * 60)
print("ONE-SAMPLE PROPORTION Z-TEST")
print("=" * 60)

print(f"Total trades             : {total_trades}")
print(f"Profitable trades        : {profitable_trades}")

print(f"Sample win rate (p̂)     : {p_hat:.4f}")
print(f"Sample win rate          : {p_hat * 100:.2f}%")

print(f"Target win rate (p₀)     : {p_0:.4f}")
print(f"Target win rate          : {p_0 * 100:.2f}%")

print(f"Standard error           : {standard_error:.4f}")
print(f"Z-statistic              : {z_stat:.4f}")
print(f"Critical Z-value         : {z_critical:.4f}")
print(f"P-value                  : {p_value:.6f}")
print(f"Significance level       : {alpha}")

print("-" * 60)

print(f"Decision using p-value   : {decision_p}")
print(f"Decision using Z-value   : {decision_z}")

print("-" * 60)

print("Conclusion:")
print(conclusion_p)

print("=" * 60)
```
