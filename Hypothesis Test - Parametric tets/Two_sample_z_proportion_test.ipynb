Below is a complete two-proportion Z-test using the trading-strategy example, including the pooled proportion, Z-statistic, p-value, critical value, and decision.

```python
import numpy as np
from scipy.stats import norm


# ============================================================
# 1. Sample data
# ============================================================

# Strategy A
n1 = 200                  # Total trades
x1 = 130                  # Profitable trades

# Strategy B
n2 = 200                  # Total trades
x2 = 110                  # Profitable trades


# ============================================================
# 2. Significance level
# ============================================================

alpha = 0.05


# ============================================================
# 3. Sample proportions
# ============================================================

p1_hat = x1 / n1
p2_hat = x2 / n2


# ============================================================
# 4. Observed difference between proportions
# ============================================================

difference = p1_hat - p2_hat


# ============================================================
# 5. Pooled proportion
#
# Used under H0: p1 = p2
# ============================================================

pooled_p = (x1 + x2) / (n1 + n2)


# ============================================================
# 6. Standard error under H0
# ============================================================

standard_error = np.sqrt(
    pooled_p * (1 - pooled_p) *
    (1 / n1 + 1 / n2)
)


# ============================================================
# 7. Z-statistic
# ============================================================

z_stat = difference / standard_error


# ============================================================
# 8. Two-tailed p-value
#
# H0: p1 = p2
# H1: p1 != p2
# ============================================================

p_value = 2 * norm.sf(abs(z_stat))


# ============================================================
# 9. Critical Z-value
# ============================================================

z_critical = norm.ppf(1 - alpha / 2)


# ============================================================
# 10. Decision using p-value
# ============================================================

if p_value < alpha:

    decision_p = "Reject H0"

    conclusion_p = (
        "There is statistically significant evidence that "
        "the win rates of Strategy A and Strategy B are different."
    )

else:

    decision_p = "Fail to reject H0"

    conclusion_p = (
        "There is insufficient statistical evidence to conclude "
        "that the win rates of Strategy A and Strategy B are different."
    )


# ============================================================
# 11. Decision using critical Z-value
# ============================================================

if abs(z_stat) > z_critical:
    decision_z = "Reject H0"
else:
    decision_z = "Fail to reject H0"


# ============================================================
# 12. Display results
# ============================================================

print("=" * 65)
print("TWO-PROPORTION Z-TEST")
print("=" * 65)

print("\n--- Sample Information ---")

print(f"Strategy A total trades       : {n1}")
print(f"Strategy A profitable trades  : {x1}")
print(f"Strategy A win rate           : {p1_hat:.4f} ({p1_hat * 100:.2f}%)")

print()

print(f"Strategy B total trades       : {n2}")
print(f"Strategy B profitable trades  : {x2}")
print(f"Strategy B win rate           : {p2_hat:.4f} ({p2_hat * 100:.2f}%)")


print("\n--- Hypotheses ---")

print("H0: p1 = p2")
print("H1: p1 != p2")


print("\n--- Test Calculation ---")

print(f"Observed difference           : {difference:.4f}")
print(f"Pooled proportion             : {pooled_p:.4f}")
print(f"Standard error                : {standard_error:.6f}")
print(f"Z-statistic                   : {z_stat:.4f}")
print(f"Critical Z-value              : ±{z_critical:.4f}")
print(f"P-value                       : {p_value:.6f}")
print(f"Significance level            : {alpha}")


print("\n--- Decision ---")

print(f"Decision using p-value        : {decision_p}")
print(f"Decision using Z-value        : {decision_z}")


print("\n--- Conclusion ---")

print(conclusion_p)

print("=" * 65)
```

Financial decision

Reject H0
: There is statistically significant evidence that Strategy A and Strategy B have different win rates at the 5% significance level.
Strategy A has the higher observed win rate:
65%−55%=10 percentage points
But this test only establishes a difference in win rate. It does not establish that Strategy A is more profitable.

If you want to test "Is Strategy A better?"

Change the test to a right-tailed test:
H0​:pA​≤pB​
H1​:pA​>pB​


```python
# Right-tailed test
p_value = norm.sf(z_stat)

z_critical = norm.ppf(1 - alpha)

if p_value < alpha:
    decision = "Reject H0"
    conclusion = (
        "There is statistically significant evidence "
        "that Strategy A has a higher win rate than Strategy B."
    )
else:
    decision = "Fail to reject H0"
    conclusion = (
        "There is insufficient evidence to conclude "
        "that Strategy A has a higher win rate than Strategy B."
    )

print(f"Z-statistic       : {z_stat:.4f}")
print(f"Critical Z        : {z_critical:.4f}")
print(f"P-value           : {p_value:.6f}")
print(f"Decision           : {decision}")
print(f"Conclusion         : {conclusion}")
```

This directly answers:

"Does Strategy A have a statistically higher win rate than Strategy B?"
