```python
import numpy as np
from scipy.stats import norm

# --------------------------------------------------
# Portfolio daily returns (%)
# --------------------------------------------------
returns = np.array([
    0.20, 0.35, 0.10, 0.45, -0.10,
    0.25, 0.15, 0.30, 0.05, 0.40,
    0.10, 0.20, 0.35, -0.05, 0.25,
    0.15, 0.30, 0.10, 0.40, 0.20,
    0.05, 0.25, 0.30, 0.15, 0.35,
    0.10, 0.20, 0.45, 0.05, 0.30
])

# --------------------------------------------------
# Hypothesized population mean return
# --------------------------------------------------
mu_0 = 0.10       # 0.10%

# Known population standard deviation
sigma = 0.50      # 0.50%

# Significance level
alpha = 0.05

# --------------------------------------------------
# Sample statistics
# --------------------------------------------------
n = len(returns)
sample_mean = np.mean(returns)

# --------------------------------------------------
# Calculate Z-statistic
# --------------------------------------------------
standard_error = sigma / np.sqrt(n)

z_stat = (sample_mean - mu_0) / standard_error

# --------------------------------------------------
# Calculate two-tailed p-value
# --------------------------------------------------
p_value = 2 * (1 - norm.cdf(abs(z_stat)))

# --------------------------------------------------
# Critical Z-value
# --------------------------------------------------
z_critical = norm.ppf(1 - alpha / 2)

# --------------------------------------------------
# Decision
# --------------------------------------------------
if p_value < alpha:
    decision = "Reject H0"
    conclusion = (
        "The portfolio's mean return is statistically "
        "significantly different from the target return."
    )
else:
    decision = "Fail to reject H0"
    conclusion = (
        "There is insufficient evidence to conclude that "
        "the portfolio's mean return is different from "
        "the target return."
    )

# --------------------------------------------------
# Results
# --------------------------------------------------
print("One-Sample Z-Test")
print("-" * 40)
print(f"Sample size          : {n}")
print(f"Sample mean return   : {sample_mean:.4f}%")
print(f"Target mean return   : {mu_0:.4f}%")
print(f"Population std (σ)   : {sigma:.4f}%")
print(f"Standard error       : {standard_error:.4f}%")
print(f"Z-statistic          : {z_stat:.4f}")
print(f"Critical Z-value     : ±{z_critical:.4f}")
print(f"P-value              : {p_value:.4f}")
print(f"Alpha                : {alpha}")
print(f"Decision              : {decision}")
print(f"Conclusion            : {conclusion}")
```

```python
import numpy as np
from scipy.stats import norm

# Portfolio returns (%)
returns = np.array([
    0.20, 0.35, 0.10, 0.45, -0.10,
    0.25, 0.15, 0.30, 0.05, 0.40,
    0.10, 0.20, 0.35, -0.05, 0.25,
    0.15, 0.30, 0.10, 0.40, 0.20,
    0.05, 0.25, 0.30, 0.15, 0.35,
    0.10, 0.20, 0.45, 0.05, 0.30
])

mu_0 = 0.10       # Target return
sigma = 0.50      # Known population standard deviation
alpha = 0.05

n = len(returns)
sample_mean = np.mean(returns)

# Standard error
se = sigma / np.sqrt(n)

# Z statistic
z = (sample_mean - mu_0) / se

# -------------------------------
# Two-tailed
# H1: μ != μ0
# -------------------------------
p_two = 2 * norm.sf(abs(z))

# -------------------------------
# Right-tailed
# H1: μ > μ0
# -------------------------------
p_greater = norm.sf(z)

# -------------------------------
# Left-tailed
# H1: μ < μ0
# -------------------------------
p_less = norm.cdf(z)

print(f"Sample mean = {sample_mean:.4f}%")
print(f"Z-statistic = {z:.4f}")

print("\nTwo-tailed test")
print(f"p-value = {p_two:.4f}")
print("Decision:",
      "Reject H0" if p_two < alpha else "Fail to reject H0")

print("\nRight-tailed test (greater)")
print(f"p-value = {p_greater:.4f}")
print("Decision:",
      "Reject H0" if p_greater < alpha else "Fail to reject H0")

print("\nLeft-tailed test (less)")
print(f"p-value = {p_less:.4f}")
print("Decision:",
      "Reject H0" if p_less < alpha else "Fail to reject H0")
```
