```python
import numpy as np
from scipy.stats import ttest_rel


def paired_t_test(before, after, alpha=0.05, alternative="two-sided"):

    # Paired t-test
    result = ttest_rel(
        after,
        before,
        alternative=alternative
    )

    t_statistic = result.statistic
    p_value = result.pvalue

    # Calculate paired differences
    differences = after - before

    mean_before = np.mean(before)
    mean_after = np.mean(after)
    mean_difference = np.mean(differences)

    # Decision
    if p_value < alpha:
        decision = "Reject H0"
    else:
        decision = "Fail to reject H0"

    # Conclusion
    if alternative == "two-sided":

        if decision == "Reject H0":
            conclusion = (
                "There is statistically significant evidence "
                "that the mean return changed after modification."
            )
        else:
            conclusion = (
                "There is insufficient statistical evidence "
                "to conclude that the mean return changed "
                "after modification."
            )

    elif alternative == "greater":

        if decision == "Reject H0":
            conclusion = (
                "There is statistically significant evidence "
                "that the modified strategy has a higher "
                "mean return than before modification."
            )
        else:
            conclusion = (
                "There is insufficient statistical evidence "
                "to conclude that the modified strategy has "
                "a higher mean return."
            )

    elif alternative == "less":

        if decision == "Reject H0":
            conclusion = (
                "There is statistically significant evidence "
                "that the modified strategy has a lower "
                "mean return than before modification."
            )
        else:
            conclusion = (
                "There is insufficient statistical evidence "
                "to conclude that the modified strategy has "
                "a lower mean return."
            )

    return {
        "n": len(before),
        "mean_before": mean_before,
        "mean_after": mean_after,
        "mean_difference": mean_difference,
        "t_statistic": t_statistic,
        "p_value": p_value,
        "alpha": alpha,
        "alternative": alternative,
        "decision": decision,
        "conclusion": conclusion
    }
```

Finance example

Suppose a trading strategy was modified, and you recorded returns for the same 10 market days before and after modification:

```python
before = np.array([
    0.10, 0.05, 0.12, -0.02, 0.08,
    0.11, 0.06, 0.09, 0.04, 0.07
])

after = np.array([
    0.15, 0.08, 0.18, 0.04, 0.11,
    0.16, 0.10, 0.14, 0.09, 0.12
])
```

Day       Before       After       Difference
------------------------------------------------
1          0.10        0.15          +0.05
2          0.05        0.08          +0.03
3          0.12        0.18          +0.06
...

Each Before observation is paired with the corresponding After observation.

1. Two-tailed test

Question:

Did the strategy's mean return change after modification?

```python
result = paired_t_test(
    before,
    after,
    alpha=0.05,
    alternative="two-sided"
)

print(result)
```

2. Right-tailed test

Question:

Did the strategy modification improve the mean return?

```python
result = paired_t_test(
    before,
    after,
    alpha=0.05,
    alternative="greater"
)

print(result)
```

3. Left-tailed test

Question:

Did the strategy modification decrease the mean return?

```python
result = paired_t_test(
    before,
    after,
    alpha=0.05,
    alternative="less"
)

print(result)
```

Complete output

You can display the result neatly:

```python
result = paired_t_test(
    before,
    after,
    alpha=0.05,
    alternative="greater"
)

print("=" * 60)
print("PAIRED T-TEST")
print("=" * 60)

print(f"Number of pairs       : {result['n']}")
print(f"Mean before           : {result['mean_before']:.4f}%")
print(f"Mean after            : {result['mean_after']:.4f}%")
print(f"Mean difference       : {result['mean_difference']:.4f}%")
print(f"T-statistic           : {result['t_statistic']:.4f}")
print(f"P-value               : {result['p_value']:.6f}")
print(f"Alpha                 : {result['alpha']}")
print(f"Alternative           : {result['alternative']}")

print("-" * 60)

print(f"Decision              : {result['decision']}")

print("-" * 60)

print("Conclusion:")
print(result["conclusion"])

print("=" * 60)
```

Decision logic

The decision is simply:

```python
if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

Key point: the paired t-test tests the mean of the paired differences, not the two means independently. This is why it is appropriate for before-vs-after strategy performance on the same dates/observations.
