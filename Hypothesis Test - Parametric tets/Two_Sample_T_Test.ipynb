Two sample T test

```python
import numpy as np
from scipy.stats import ttest_ind


# ============================================================
# Sample data
# Daily returns (%) from two independent strategies
# ============================================================

strategy_A = np.array([
    0.12, 0.18, 0.05, 0.22, 0.10,
    0.15, 0.08, 0.20, 0.14, 0.11,
    0.17, 0.09, 0.13, 0.16, 0.07,
    0.19, 0.12, 0.21, 0.10, 0.15
])

strategy_B = np.array([
    0.08, 0.12, 0.04, 0.15, 0.09,
    0.11, 0.06, 0.14, 0.10, 0.07,
    0.13, 0.05, 0.09, 0.12, 0.06,
    0.14, 0.08, 0.16, 0.07, 0.11
])

alpha = 0.05


# ============================================================
# Welch's Two-Sample T-Test
# ============================================================

def welch_t_test(strategy_A, strategy_B,
                 alpha=0.05,
                 alternative="two-sided"):

    # Welch's t-test
    result = ttest_ind(
        strategy_A,
        strategy_B,
        equal_var=False,       # IMPORTANT: Welch's t-test
        alternative=alternative
    )

    t_statistic = result.statistic
    p_value = result.pvalue

    # Sample statistics
    mean_A = np.mean(strategy_A)
    mean_B = np.mean(strategy_B)

    std_A = np.std(strategy_A, ddof=1)
    std_B = np.std(strategy_B, ddof=1)

    n_A = len(strategy_A)
    n_B = len(strategy_B)

    mean_difference = mean_A - mean_B

    # Decision
    if p_value < alpha:
        decision = "Reject H0"
    else:
        decision = "Fail to reject H0"

    # --------------------------------------------------------
    # Conclusion based on alternative
    # --------------------------------------------------------

    if alternative == "two-sided":

        if decision == "Reject H0":
            conclusion = (
                "There is statistically significant evidence "
                "that the mean returns of Strategy A and "
                "Strategy B are different."
            )
        else:
            conclusion = (
                "There is insufficient statistical evidence "
                "to conclude that the mean returns of Strategy A "
                "and Strategy B are different."
            )

    elif alternative == "greater":

        if decision == "Reject H0":
            conclusion = (
                "There is statistically significant evidence "
                "that Strategy A has a higher mean return "
                "than Strategy B."
            )
        else:
            conclusion = (
                "There is insufficient statistical evidence "
                "to conclude that Strategy A has a higher mean "
                "return than Strategy B."
            )

    elif alternative == "less":

        if decision == "Reject H0":
            conclusion = (
                "There is statistically significant evidence "
                "that Strategy A has a lower mean return "
                "than Strategy B."
            )
        else:
            conclusion = (
                "There is insufficient statistical evidence "
                "to conclude that Strategy A has a lower mean "
                "return than Strategy B."
            )

    return {
        "n_A": n_A,
        "n_B": n_B,
        "mean_A": mean_A,
        "mean_B": mean_B,
        "std_A": std_A,
        "std_B": std_B,
        "mean_difference": mean_difference,
        "t_statistic": t_statistic,
        "p_value": p_value,
        "alpha": alpha,
        "alternative": alternative,
        "decision": decision,
        "conclusion": conclusion
    }


# ============================================================
# TWO-TAILED TEST
# H0: μA = μB
# H1: μA != μB
# ============================================================

result = welch_t_test(
    strategy_A,
    strategy_B,
    alpha=0.05,
    alternative="two-sided"
)


# ============================================================
# Display results
# ============================================================

print("=" * 65)
print("WELCH'S TWO-SAMPLE T-TEST")
print("=" * 65)

print("\nSample Information")
print("-" * 65)

print(f"Strategy A observations : {result['n_A']}")
print(f"Strategy B observations : {result['n_B']}")

print(f"Strategy A mean        : {result['mean_A']:.4f}%")
print(f"Strategy B mean        : {result['mean_B']:.4f}%")

print(f"Strategy A std         : {result['std_A']:.4f}%")
print(f"Strategy B std         : {result['std_B']:.4f}%")

print(f"Mean difference (A-B)  : {result['mean_difference']:.4f}%")

print("\nHypothesis Test")
print("-" * 65)

print(f"Alternative             : {result['alternative']}")
print(f"t-statistic             : {result['t_statistic']:.4f}")
print(f"p-value                 : {result['p_value']:.6f}")
print(f"Significance level      : {result['alpha']}")

print("\nDecision")
print("-" * 65)

print(f"Decision                : {result['decision']}")

print("\nConclusion")
print("-" * 65)

print(result["conclusion"])

print("=" * 65)
```

The three alternatives
1. Two-tailed — "Are they different?"

```python
result = welch_t_test(
    strategy_A,
    strategy_B,
    alpha=0.05,
    alternative="two-sided"
)
```

2. Right-tailed — "Is Strategy A better?"

If your question is:

Does Strategy A have a higher average return than Strategy B?

Use:

```python
result = welch_t_test(
    strategy_A,
    strategy_B,
    alpha=0.05,
    alternative="greater"
)
```

3. Left-tailed — "Is Strategy A worse?"

If your question is:

Does Strategy A have a lower average return than Strategy B?

Use:

```python
result = welch_t_test(
    strategy_A,
    strategy_B,
    alpha=0.05,
    alternative="less"
)
```

Why equal_var=False?

This is the important part:

```python
ttest_ind(
    strategy_A,
    strategy_B,
    equal_var=False,
    alternative=alternative
)
```

equal_var=False tells SciPy to perform Welch's t-test, which does not assume that Strategy A and Strategy B have equal population variances.

For financial returns, this is generally a good default because two strategies can have very different volatility.
