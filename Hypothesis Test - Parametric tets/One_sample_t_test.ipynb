**One sample T Test**

```python
import numpy as np
from scipy.stats import ttest_1samp


# ============================================================
# Portfolio daily returns
# ============================================================

portfolio_returns = np.array([
    0.12, 0.18, 0.05, 0.22, 0.10,
    0.15, 0.08, 0.20, 0.14, 0.11,
    0.17, 0.09, 0.13, 0.16, 0.07,
    0.19, 0.12, 0.21, 0.10, 0.15,
    0.06, 0.18, 0.14, 0.11, 0.20,
    0.09, 0.16, 0.13, 0.17, 0.12
])

# Target / hypothesized population mean
target_return = 0.10

# Significance level
alpha = 0.05


# ============================================================
# Function for One-Sample T-Test
# ============================================================

def one_sample_t_test(data, population_mean, alpha=0.05,
                      alternative="two-sided"):

    # Perform one-sample t-test
    result = ttest_1samp(
        data,
        popmean=population_mean,
        alternative=alternative
    )

    t_statistic = result.statistic
    p_value = result.pvalue

    # Sample statistics
    sample_size = len(data)
    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)

    # Decision
    if p_value < alpha:
        decision = "Reject H0"
    else:
        decision = "Fail to reject H0"

    # Conclusion
    if alternative == "two-sided":

        if decision == "Reject H0":
            conclusion = (
                "There is statistically significant evidence that "
                "the portfolio mean return is different from "
                f"{population_mean:.2f}%."
            )
        else:
            conclusion = (
                "There is insufficient statistical evidence to "
                "conclude that the portfolio mean return is "
                f"different from {population_mean:.2f}%."
            )

    elif alternative == "greater":

        if decision == "Reject H0":
            conclusion = (
                "There is statistically significant evidence that "
                "the portfolio mean return is greater than "
                f"{population_mean:.2f}%."
            )
        else:
            conclusion = (
                "There is insufficient statistical evidence to "
                "conclude that the portfolio mean return is greater "
                f"than {population_mean:.2f}%."
            )

    elif alternative == "less":

        if decision == "Reject H0":
            conclusion = (
                "There is statistically significant evidence that "
                "the portfolio mean return is less than "
                f"{population_mean:.2f}%."
            )
        else:
            conclusion = (
                "There is insufficient statistical evidence to "
                "conclude that the portfolio mean return is less "
                f"than {population_mean:.2f}%."
            )

    return {
        "sample_size": sample_size,
        "sample_mean": sample_mean,
        "sample_std": sample_std,
        "target_mean": population_mean,
        "t_statistic": t_statistic,
        "p_value": p_value,
        "alpha": alpha,
        "alternative": alternative,
        "decision": decision,
        "conclusion": conclusion
    }


# ============================================================
# TWO-TAILED TEST
# Is mean return different from 0.10%?
# ============================================================

result = one_sample_t_test(
    portfolio_returns,
    population_mean=target_return,
    alpha=alpha,
    alternative="two-sided"
)


# ============================================================
# Display results
# ============================================================

print("=" * 60)
print("ONE-SAMPLE T-TEST")
print("=" * 60)

print(f"Sample size       : {result['sample_size']}")
print(f"Sample mean       : {result['sample_mean']:.4f}%")
print(f"Sample std        : {result['sample_std']:.4f}%")
print(f"Target mean       : {result['target_mean']:.4f}%")
print(f"T-statistic       : {result['t_statistic']:.4f}")
print(f"P-value           : {result['p_value']:.6f}")
print(f"Alpha             : {result['alpha']}")
print(f"Alternative       : {result['alternative']}")

print("-" * 60)

print(f"Decision          : {result['decision']}")

print("-" * 60)

print("Conclusion:")
print(result["conclusion"])

print("=" * 60)
```

The three alternatives

You only need to change the alternative parameter.

1. Two-tailed

Is the portfolio mean return different from 0.10%?

```python
result = ttest_1samp(
    portfolio_returns,
    popmean=0.10,
    alternative="two-sided"
)
```

2. Right-tailed

Is the portfolio mean return greater than 0.10%?

```python
result = ttest_1samp(
    portfolio_returns,
    popmean=0.10,
    alternative="greater"
)
```

3. Left-tailed

Is the portfolio mean return less than 0.10%?

```python
result = ttest_1samp(
    portfolio_returns,
    popmean=0.10,
    alternative="less"
)
```

Decision logic

The decision-making part remains very simple:

```python
if result.pvalue < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

So the overall pattern is:

portfolio returns
       ↓
scipy.stats.ttest_1samp()
       ↓
t-statistic + p-value
       ↓
Compare p-value with α
       ↓
   ┌───────────────┐
   │ p < α         │
   │ Reject H0     │
   └───────────────┘

   ┌───────────────┐
   │ p ≥ α         │
   │ Fail to reject│
   │ H0            │
   └───────────────┘
