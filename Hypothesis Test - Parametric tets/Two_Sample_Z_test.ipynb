Below is a complete two-sample Z-test for comparing the mean returns of two independent investment strategies, including the Z-statistic, p-value, critical value, and decision.

```python
import numpy as np
from scipy.stats import norm

# ============================================================
# 1. Sample data
# ============================================================

# Daily returns of Strategy A (%)
strategy_a = np.array([
    0.20, 0.35, 0.10, 0.45, -0.10,
    0.25, 0.15, 0.30, 0.05, 0.40,
    0.10, 0.20, 0.35, -0.05, 0.25,
    0.15, 0.30, 0.10, 0.40, 0.20,
    0.05, 0.25, 0.30, 0.15, 0.35,
    0.10, 0.20, 0.45, 0.05, 0.30
])

# Daily returns of Strategy B (%)
strategy_b = np.array([
    0.10, 0.20, 0.05, 0.30, -0.05,
    0.15, 0.10, 0.20, 0.00, 0.25,
    0.05, 0.15, 0.25, -0.10, 0.20,
    0.10, 0.20, 0.05, 0.30, 0.15,
    0.00, 0.20, 0.25, 0.10, 0.20,
    0.05, 0.15, 0.30, 0.00, 0.20
])


# ============================================================
# 2. Known population standard deviations
# ============================================================

sigma_a = 0.50    # Known population SD of Strategy A (%)
sigma_b = 0.40    # Known population SD of Strategy B (%)


# ============================================================
# 3. Significance level
# ============================================================

alpha = 0.05


# ============================================================
# 4. Sample statistics
# ============================================================

n_a = len(strategy_a)
n_b = len(strategy_b)

mean_a = np.mean(strategy_a)
mean_b = np.mean(strategy_b)


# ============================================================
# 5. Difference between sample means
# ============================================================

mean_difference = mean_a - mean_b


# ============================================================
# 6. Standard error of difference
# ============================================================

standard_error = np.sqrt(
    (sigma_a ** 2 / n_a) +
    (sigma_b ** 2 / n_b)
)


# ============================================================
# 7. Calculate Z-statistic
# ============================================================

z_stat = mean_difference / standard_error


# ============================================================
# 8. Two-tailed p-value
# H1: μA != μB
# ============================================================

p_value = 2 * norm.sf(abs(z_stat))


# ============================================================
# 9. Critical Z-value
# ============================================================

z_critical = norm.ppf(1 - alpha / 2)


# ============================================================
# 10. Decision
# ============================================================

if p_value < alpha:
    decision = "Reject H0"
    conclusion = (
        "There is statistically significant evidence that "
        "the mean returns of Strategy A and Strategy B are different."
    )
else:
    decision = "Fail to reject H0"
    conclusion = (
        "There is insufficient statistical evidence to conclude "
        "that the mean returns of Strategy A and Strategy B are different."
    )


# ============================================================
# 11. Display results
# ============================================================

print("=" * 60)
print("TWO-SAMPLE Z-TEST")
print("=" * 60)

print(f"Strategy A sample size       : {n_a}")
print(f"Strategy B sample size       : {n_b}")

print(f"Strategy A mean return       : {mean_a:.4f}%")
print(f"Strategy B mean return       : {mean_b:.4f}%")

print(f"Strategy A population SD     : {sigma_a:.4f}%")
print(f"Strategy B population SD     : {sigma_b:.4f}%")

print(f"Mean difference (A - B)      : {mean_difference:.4f}%")
print(f"Standard error               : {standard_error:.4f}%")

print(f"Z-statistic                  : {z_stat:.4f}")
print(f"Critical Z-values            : ±{z_critical:.4f}")
print(f"P-value                      : {p_value:.6f}")
print(f"Significance level           : {alpha}")

print("-" * 60)
print(f"Decision                     : {decision}")
print(f"Conclusion                   : {conclusion}")
print("=" * 60)
```

Decision using p-value

The code uses:

```python
if p_value < alpha:
    decision = "Reject H0"
else:
    decision = "Fail to reject H0"
```

```python
alpha = 0.05
```

the rule is:
p<0.05⇒Reject H0​​
p≥0.05⇒Fail to reject H0​​

Decision using Z-statistic

You can also make the decision directly from the Z-statistic:

```python
if abs(z_stat) > z_critical:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

If your question is "Is Strategy A better?"

Then you should use a right-tailed test instead:
H0​:μA​≤μB​
H1​:μA​>μB​
Change the p-value and critical value to:

```python
# Right-tailed test
p_value = norm.sf(z_stat)

z_critical = norm.ppf(1 - alpha)

if p_value < alpha:
    decision = "Reject H0"
    conclusion = (
        "There is statistically significant evidence that "
        "Strategy A has a higher mean return than Strategy B."
    )
else:
    decision = "Fail to reject H0"
    conclusion = (
        "There is insufficient evidence to conclude that "
        "Strategy A has a higher mean return than Strategy B."
    )
```

Here the critical value is:
Zcritical​=1.645
So:
Z>1.645⇒Strategy A significantly outperforms B​
Important: In real financial data, you generally don't know the true population standard deviations std1, std2.If you estimate them from the samples, use Welch's two-sample t-test rather than this Z-test.
