**Standard deviation of returns**

```python
import yfinance as yf
import pandas as pd
from datetime import datetime, timedelta

# Define the ticker symbol for TCS (Tata Consultancy Services)
ticker_symbol = "TCS.NS"  # .NS suffix for NSE, India

# Calculate the date range for one year
end_date = datetime.now()
start_date = end_date - timedelta(days=365)

# Download historical data for TCS
df = yf.download(ticker_symbol, start=start_date, end=end_date)

# Calculate daily returns
df["Daily_Return"] = df["Close"].pct_change()

# Create a new DataFrame with only daily returns (and drop any NaN from the first row)
daily_returns_df = df[["Daily_Return"]].dropna()

print(f"Downloaded daily returns for {ticker_symbol} for the last year:")
print(daily_returns_df.head())
```

```python
import pandas as pd
import numpy as np

# Example:
# df["Daily_Return"] = df["Close"].pct_change()

# Remove missing values
returns = df["Daily_Return"].dropna()

# Sample Standard Deviation (Volatility)
#volatility = returns.std(ddof=1)
volatility = np.std(returns,ddof=1)
print("=" * 60)
print(f"Standard Deviation (Volatility): {volatility:.4%}")
print("=" * 60)

# Decision Making
if volatility < 0.01:
    print("Volatility Level : Very Low")
    print("Risk             : Very Low")
    print("Interpretation   : Price movements are very stable.")
    print("Suitable For     : Conservative investors")
    print("Decision         : Can be considered for capital preservation.")

elif 0.01 <= volatility < 0.02:
    print("Volatility Level : Low")
    print("Risk             : Low")
    print("Interpretation   : Moderate day-to-day price fluctuations.")
    print("Suitable For     : Long-term investors")
    print("Decision         : Suitable for diversified portfolios.")

elif 0.02 <= volatility < 0.04:
    print("Volatility Level : Moderate")
    print("Risk             : Medium")
    print("Interpretation   : Returns fluctuate noticeably.")
    print("Suitable For     : Balanced investors")
    print("Decision         : Position sizing and diversification recommended.")

elif 0.04 <= volatility < 0.06:
    print("Volatility Level : High")
    print("Risk             : High")
    print("Interpretation   : Large price swings are common.")
    print("Suitable For     : Active traders")
    print("Decision         : Use stop-loss and monitor positions closely.")

else:
    print("Volatility Level : Very High")
    print("Risk             : Very High")
    print("Interpretation   : Extreme price movements.")
    print("Suitable For     : Experienced traders only")
    print("Decision         :")
    print("  • Reduce position size")
    print("  • Diversify the portfolio")
    print("  • Consider hedging with options or futures")
    print("  • Perform stress testing")
```
