# Chi-Square Test of Independence with Decision Making

This notebook demonstrates how to perform a **Chi-square test of independence** using `scipy.stats.chi2_contingency()`.

### Example

We compare:

- **Trading Strategy**: Strategy A vs Strategy B
- **Trade Outcome**: Win vs Loss

The question is:

> Is trading strategy associated with trade outcome?

### Hypotheses

**H₀:** Trading Strategy and Trade Outcome are independent.

**H₁:** Trading Strategy and Trade Outcome are associated.

### Decision rule

Using α = 0.05:

- **p-value < 0.05** → Reject H₀
- **p-value ≥ 0.05** → Fail to reject H₀


```python
# Import libraries

import numpy as np
import pandas as pd

from scipy.stats import chi2_contingency
```

## 1. Create the observed frequency table

Suppose we observed the following trading results:

| Strategy | Win | Loss |
|---|---:|---:|
| Strategy A | 65 | 35 |
| Strategy B | 55 | 45 |

The rows represent the trading strategies and the columns represent the trade outcomes.


```python
# Observed frequency table

observed = np.array([
    [65, 35],   # Strategy A
    [55, 45]    # Strategy B
])

contingency_table = pd.DataFrame(
    observed,
    index=["Strategy A", "Strategy B"],
    columns=["Win", "Loss"]
)

print("Observed contingency table:")
print(contingency_table)
```

## 2. Perform the Chi-square test

`chi2_contingency()` returns:

- `chi2`: Chi-square test statistic
- `p_value`: p-value
- `degrees_of_freedom`: degrees of freedom
- `expected`: expected frequencies under H₀


```python
# Perform Chi-square test of independence

chi2, p_value, degrees_of_freedom, expected = chi2_contingency(
    observed
)

print("Chi-square statistic :", chi2)
print("p-value              :", p_value)
print("Degrees of freedom   :", degrees_of_freedom)
```

## 3. Display the expected frequencies

The expected frequency represents what we would expect to observe if the two categorical variables were independent.

Formula:

**Expected frequency = (row total × column total) / grand total**


```python
expected_table = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

print("Expected frequencies under H₀:")
print(expected_table)
```

## 4. Make the statistical decision

Use:

**α = 0.05**

Decision:

- If `p-value < α` → **Reject H₀**
- If `p-value >= α` → **Fail to reject H₀**


```python
# Decision making

alpha = 0.05

if p_value < alpha:
    decision = "Reject H0"
    conclusion = (
        "There is statistically significant evidence that "
        "Trading Strategy and Trade Outcome are associated."
    )
else:
    decision = "Fail to reject H0"
    conclusion = (
        "There is insufficient statistical evidence to conclude "
        "that Trading Strategy and Trade Outcome are associated."
    )

print("Significance level (alpha):", alpha)
print("p-value                  :", p_value)
print("Decision                 :", decision)
print("Conclusion               :", conclusion)
```

## 5. Complete test function

The following function can be reused for any categorical contingency table.


```python
def chi_square_independence_test(
    observed,
    alpha=0.05,
    row_labels=None,
    column_labels=None
):
    """Perform a Chi-square test of independence and make a decision."""

    chi2, p_value, dof, expected = chi2_contingency(observed)

    if p_value < alpha:
        decision = "Reject H0"
        conclusion = (
            "There is statistically significant evidence of an "
            "association between the two categorical variables."
        )
    else:
        decision = "Fail to reject H0"
        conclusion = (
            "There is insufficient statistical evidence to conclude "
            "that the two categorical variables are associated."
        )

    if row_labels is not None and column_labels is not None:
        observed_df = pd.DataFrame(
            observed,
            index=row_labels,
            columns=column_labels
        )

        expected_df = pd.DataFrame(
            expected,
            index=row_labels,
            columns=column_labels
        )
    else:
        observed_df = pd.DataFrame(observed)
        expected_df = pd.DataFrame(expected)

    return {
        "observed": observed_df,
        "expected": expected_df,
        "chi2": chi2,
        "p_value": p_value,
        "degrees_of_freedom": dof,
        "alpha": alpha,
        "decision": decision,
        "conclusion": conclusion
    }
```

## 6. Run the reusable function

```python
result = chi_square_independence_test(
    observed=observed,
    alpha=0.05,
    row_labels=["Strategy A", "Strategy B"],
    column_labels=["Win", "Loss"]
)

print("Observed:")
print(result["observed"])

print("\nExpected:")
print(result["expected"])

print("\nChi-square statistic:", result["chi2"])
print("p-value:", result["p_value"])
print("Degrees of freedom:", result["degrees_of_freedom"])
print("Alpha:", result["alpha"])
print("Decision:", result["decision"])
print("Conclusion:", result["conclusion"])
```

## 7. Finance interpretation

The test does **not** directly test whether Strategy A is more profitable.

It tests whether:

**Strategy choice** and **trade outcome (Win/Loss)** are statistically associated.

If H₀ is rejected:

> There is statistical evidence that trade outcomes are associated with the choice of strategy.

If H₀ is not rejected:

> There is insufficient statistical evidence to conclude that trade outcomes are associated with the choice of strategy.

To determine whether Strategy A is actually better, also examine:

- Average return
- Volatility
- Sharpe ratio
- Maximum drawdown
- Transaction costs
- Out-of-sample performance


## 8. Important assumptions

The Chi-square test of independence is appropriate when:

1. Both variables are categorical.
2. Observations are appropriately independent.
3. Data are represented as frequency/counts.
4. Expected frequencies are not too small.

A common practical guideline is that expected cell counts should generally be at least about 5.

If expected counts are too small, consider an exact test such as **Fisher's exact test** for appropriate 2×2 tables.


```python
# Check the minimum expected frequency

minimum_expected = expected.min()

print("Minimum expected frequency:", minimum_expected)

if minimum_expected >= 5:
    print("Expected-frequency check: PASS")
else:
    print("Expected-frequency check: REVIEW")
```

**H₀:**

> The variables are associated.

**Decision at α = 0.05:**

- `p-value < 0.05` → Reject H₀

The `scipy.stats.chi2_contingency()` function performs the test and provides the test statistic, p-value, degrees of freedom, and expected frequencies.
