**Wilcoxcon signed rank test for before vs after modifying the same trading strategy, the returns are naturally paired by trading day.**

```python
import numpy as np
from scipy.stats import wilcoxon


# ============================================================
# 1. Example: Strategy returns BEFORE modification
# ============================================================

before = np.array([
    0.010, 0.008, 0.012, -0.004, 0.006,
    0.009, -0.002, 0.011, 0.007, 0.005,
    0.013, 0.004, 0.008, -0.003, 0.010,
    0.006, 0.009, 0.002, 0.011, 0.007
])


# ============================================================
# 2. Example: Strategy returns AFTER modification
#
# Same trading days as "before"
# ============================================================

after = np.array([
    0.013, 0.010, 0.015, -0.001, 0.009,
    0.011,  0.001, 0.014, 0.009, 0.007,
    0.016, 0.006, 0.011, 0.000, 0.013,
    0.009, 0.012, 0.004, 0.014, 0.010
])


# ============================================================
# 3. Difference
# ============================================================

difference = after - before

print("Paired differences (After - Before):")
print(difference)

print("\nAverage difference :", np.mean(difference))
print("Median difference  :", np.median(difference))


# ============================================================
# 4. Function to perform Wilcoxon test
# ============================================================

def wilcoxon_test(before, after, alternative, alpha=0.05):

    result = wilcoxon(
        after,
        before,
        alternative=alternative
    )

    print("\n" + "=" * 65)

    if alternative == "two-sided":
        print("WILCOXON SIGNED-RANK TEST: TWO-SIDED")

    elif alternative == "greater":
        print("WILCOXON SIGNED-RANK TEST: GREATER")

    elif alternative == "less":
        print("WILCOXON SIGNED-RANK TEST: LESS")

    print("=" * 65)

    print(f"Wilcoxon statistic : {result.statistic}")
    print(f"p-value            : {result.pvalue}")
    print(f"alpha              : {alpha}")

    # --------------------------------------------------------
    # Decision
    # --------------------------------------------------------

    if result.pvalue <= alpha:

        print("\nDecision: REJECT H0")

        if alternative == "two-sided":
            print(
                "There is statistically significant evidence "
                "that the strategy returns changed after "
                "the modification."
            )

        elif alternative == "greater":
            print(
                "There is statistically significant evidence "
                "that returns AFTER the modification are "
                "generally higher than BEFORE."
            )

        elif alternative == "less":
            print(
                "There is statistically significant evidence "
                "that returns AFTER the modification are "
                "generally lower than BEFORE."
            )

    else:

        print("\nDecision: FAIL TO REJECT H0")

        if alternative == "two-sided":
            print(
                "There is insufficient statistical evidence "
                "that the strategy returns changed after "
                "the modification."
            )

        elif alternative == "greater":
            print(
                "There is insufficient statistical evidence "
                "that returns AFTER the modification are "
                "generally higher than BEFORE."
            )

        elif alternative == "less":
            print(
                "There is insufficient statistical evidence "
                "that returns AFTER the modification are "
                "generally lower than BEFORE."
            )

    return result


# ============================================================
# 5. TWO-SIDED TEST
#
# Question:
# Did the strategy change in either direction?
#
# H0: No systematic change
# H1: There is a systematic change
# ============================================================

two_sided = wilcoxon_test(
    before,
    after,
    alternative="two-sided"
)


# ============================================================
# 6. GREATER TEST
#
# Question:
# Did the strategy improve?
#
# H0: After <= Before
# H1: After > Before
# ============================================================

greater = wilcoxon_test(
    before,
    after,
    alternative="greater"
)


# ============================================================
# 7. LESS TEST
#
# Question:
# Did the strategy become worse?
#
# H0: After >= Before
# H1: After < Before
# ============================================================

less = wilcoxon_test(
    before,
    after,
    alternative="less"
)
```
